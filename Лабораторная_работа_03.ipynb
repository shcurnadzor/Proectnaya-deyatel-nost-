## Задание

Напишите функцию, которая печатает таблицу значений функции $$y=𝑥^2+3 \cdot 𝑥^3$$ для 𝑥 ∈ [0,10]. Таблица печатает значения только
для целых значений 𝑥 с шагом 1

## Решение

### Объявление функции

```python
def f(x, y):
  s = 0
  for i in range(x,y):
    s+=i**2+3*i**3
  return s
```

### Демонстрация использования функции

```python
x, y = 1, 2
print(f'Сумма квадратов числел от {x} до {y}: {f(x, y)}')
x, y = 500, 50000
print(f'Сумма квадратов числел от {x} до {y}: {f(x, y)}')
x, y = 268, 3258
print(f'Сумма квадратов числел от {x} до {y}: {f(x, y)}')
```
